```python
def procesar_ventas(ventas):

    total_ventas = 0
    cantidad_ventas = 0

    for venta in ventas:

        precio = venta["precio"]
        cantidad = venta["cantidad"]

        subtotal = precio * cantidad

        total_ventas += subtotal
        cantidad_ventas += 1

    promedio = total_ventas / cantidad_ventas

    resultado = {
        "total_ventas": total_ventas,
        "cantidad_ventas": cantidad_ventas,
        "promedio": promedio
    }

    return resultado


# Datos de prueba
ventas = [
    {"precio": 100, "cantidad": 2},
    {"precio": 50, "cantidad": 3},
    {"precio": 80, "cantidad": 1}
]

resultado = procesar_ventas(ventas)

print(resultado)
```
