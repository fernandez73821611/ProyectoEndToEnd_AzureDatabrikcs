# Magic Commands en Azure Databricks

## ¿Qué son?

Los Magic Commands son comandos especiales que permiten realizar diferentes operaciones dentro de los notebooks de Azure Databricks. Se identifican porque comienzan con el símbolo `%`.

## Principales Magic Commands

| Comando | ¿Para qué sirve? |
|---|---|
| `%python` | Ejecutar código Python y PySpark. |
| `%sql` | Ejecutar consultas SQL. |
| `%scala` | Ejecutar código Scala. |
| `%r` | Ejecutar código en lenguaje R. |
| `%md` | Agregar texto y documentación en formato Markdown. |
| `%pip` | Instalar librerías de Python. |
| `%run` | Ejecutar otro notebook y reutilizar su código. |
| `%fs` | Explorar y administrar archivos y directorios. |
| `%sh` | Ejecutar comandos de Shell (Linux). |
| `%lsmagic` | Consultar los comandos mágicos disponibles en el entorno. |
