# Leer datos JSON de la carpeta Employees
- Consultar archivo Employees mediate formato JSON
- Consultar archivo Employees mediate formato TEXT
- Crear una view en el schema bronze

## 01. Consultar archivo Employees mediate formato JSON

```sql
CREATE OR REPLACE TEMPORARY VIEW tv_employees 
USING JSON
OPTIONS(
    path '/Volumes/zenviro/raw/operational_data/employees/'
)
```

```sql
SELECT *FROM tv_employees limit 10
```

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/employees/',
    format=>'json'
)Limit 10
```

## 02. Consultar archivo Employees mediate formato TEXT

```sql
CREATE OR REPLACE TEMPORARY VIEW tv_employees
USING TEXT
OPTIONS(
    path '/Volumes/zenviro/raw/operational_data/employees/'
)
```

```sql
SELECT *FROM tv_employees limit 10
```

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/employees/',
    format=>'text'
)limit 10
```

## 03. Crear una view en el schema bronze

```sql
CREATE OR REPLACE VIEW zenviro.bronze.v_employees
AS
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/employees/',
    format=>'text'
)
```

```sql
SELECT *FROM zenviro.bronze.v_employees LIMIT 10
```
