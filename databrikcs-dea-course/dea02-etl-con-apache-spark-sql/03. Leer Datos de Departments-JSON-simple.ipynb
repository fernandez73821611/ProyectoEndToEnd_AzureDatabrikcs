# Leer datos de Deparments JSON simple
- Leer archivos departments
- Agregar el metadato file_path
- Crear un vista y guardar en schema bronze

## 01. Leer archivos departments

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/departments/',
    format=>'json'
)Limit 10;
```

## 02. Agregar el metadato file_path

```sql
SELECT _metadata.file_path,*FROM read_files(
    '/Volumes/zenviro/raw/operational_data/departments/',
    format=>'json'
)
```

## 03. Crear un vista y guardar en schema bronze

```sql
CREATE OR REPLACE VIEW zenviro.bronze.v_departments
AS
SELECT _metadata.file_path,*FROM read_files(
    '/Volumes/zenviro/raw/operational_data/departments/',
    format=>'json'
)
```

```sql
SELECT *FROM zenviro.bronze.v_departments limit 10;
```
