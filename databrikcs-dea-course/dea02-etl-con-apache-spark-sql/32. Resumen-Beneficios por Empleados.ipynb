#  Resumen-Beneficios por Empleados

```sql
SELECT *FROM zenviro.silver.benefits limit 5
```

```sql
SELECT *FROM zenviro.silver.employee_benefits limit 5
```

```sql
SELECT *FROM zenviro.silver.employees limit 5
```

```sql
SELECT 
b.benefit_name,
COUNT(e.employee_id) as total_employees
    FROM zenviro.silver.benefits  b 
    INNER JOIN zenviro.silver.employee_benefits eb
    ON b.benefit_id = eb.benefit_id
    INNER JOIN zenviro.silver.employees e
    ON eb.employee_id = e.employee_id
    WHERE year(eb.start_date) IN (2023,2024) AND eb.end_date IS NULL

GROUP BY b.benefit_name
ORDER BY COUNT(e.employee_id) DESC;
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.bronze.employee_benefits_summary
AS
SELECT 
b.benefit_name,
COUNT(e.employee_id) as total_employees
    FROM zenviro.silver.benefits  b 
    INNER JOIN zenviro.silver.employee_benefits eb
    ON b.benefit_id = eb.benefit_id
    INNER JOIN zenviro.silver.employees e
    ON eb.employee_id = e.employee_id
    WHERE year(eb.start_date) IN (2023,2024) AND eb.end_date IS NULL

GROUP BY b.benefit_name
ORDER BY COUNT(e.employee_id) DESC;
```
