# Creacion de Temporary Views

Las vistas temporales son efimeras y no son obejetos en Unity Catalog, se utilizan cuando tienes un ETL complejo

- Creacion de un temporary view: solo permanece en la session
- Creacion de un global temporary view: permanece a nivel de cluster y puede ser utilizado por mas noteboocks del mismo cluster

## 01. Creacion de un temporary view

```sql
CREATE OR REPLACE TEMPORARY VIEW tv_locations
AS
SELECT _metadata.file_path,*FROM read_files(
    '/Volumes/zenviro/raw/operational_data/locations/'
)
```

```sql
SELECT *FROM tv_locations limit 10;
```

## 02. Creacion de un global temporary view

```sql
CREATE OR REPLACE GLOBAL TEMPORARY VIEW gtv_locations
AS
SELECT _metadata.file_path,*FROM read_files(
    '/Volumes/zenviro/raw/operational_data/locations/'
)
```

```sql
SELECT *FROM global_temp.gtv_locations limit 10;
```
