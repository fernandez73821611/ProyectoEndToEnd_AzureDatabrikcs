# Transformar Datos Benefits
- Sustituir el caracter "-" por "espacio" en el campo descripcion
- Escribir los datos transformados  en el schema silver

```sql
SELECT *FROM zenviro.bronze.benefits
```

```sql
SELECT 
benefit_id,
benefit_name,
replace(description, '-', ' ') AS description,
type
FROM zenviro.bronze.benefits
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.silver.benefits
AS
SELECT 
benefit_id,
benefit_name,
replace(description, '-', ' ') AS description,
type
FROM zenviro.bronze.benefits
```
