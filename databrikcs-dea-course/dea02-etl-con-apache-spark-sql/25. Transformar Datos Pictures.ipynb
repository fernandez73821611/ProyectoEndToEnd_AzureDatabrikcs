#  Transformar Datos Pictures
- Obtener el location_id del path del archivo
- Escribir los datos transformados en el schema silver

```sql
SELECT *FROM zenviro.bronze.v_pictures limit 5
```

```sql
SELECT 
CAST( regexp_extract(path,'.*/([0-9]+)\\.jpg$',1) AS INT) as location_id,
content as location_picture
FROM zenviro.bronze.v_pictures
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.silver.picture
AS
SELECT 
CAST( regexp_extract(path,'.*/([0-9]+)\\.jpg$',1) AS INT) as location_id,
content as location_picture
FROM zenviro.bronze.v_pictures
```
