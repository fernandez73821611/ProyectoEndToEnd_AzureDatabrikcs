# Leer Datos Attendances Archivos TSV

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/attendances/',
     format => 'csv',
     sep=>'\t',
     header=>'true'
)limit 5;
```

```sql
CREATE OR REPLACE VIEW zenviro.bronze.v_attendances
AS
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/attendances/',
     format => 'csv',
     sep=>'\t',
     header=>'true'
)
```

```sql
SELECT *FROM zenviro.bronze.v_attendances limit 10
```
