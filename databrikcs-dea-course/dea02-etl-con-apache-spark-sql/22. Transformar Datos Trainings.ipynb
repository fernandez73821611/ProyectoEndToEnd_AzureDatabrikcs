# Transformar Datos Trainings
- Converir solo la primera letra en mayuscula de los campos training_name, description y provider
- Escribir los datos transformados en silver

```sql
SELECT *FROM zenviro.bronze.trainings
```

```sql
SELECT 
training_id,
INITCAP(training_name) AS training_name,
INITCAP(description) AS description,
INITCAP(provider) AS provider,
duration_hours
FROM zenviro.bronze.trainings
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.silver.trainings
AS
SELECT 
training_id,
INITCAP(training_name) AS training_name,
INITCAP(description) AS description,
INITCAP(provider) AS provider,
duration_hours
FROM zenviro.bronze.trainings
```
