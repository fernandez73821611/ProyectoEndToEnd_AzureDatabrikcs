# Transformar Datos Employee Training
- Extrae la fecha y hora de completition_date en dos columnas completition_date, time
- Asignar al campo status los siguientes valores: 1: Not Started, 2: In Progress 3:Completed
- Escribir los datos transformados en el schema silver

```sql
SELECT *FROM zenviro.bronze.employee_training
```

```sql
SELECT 
employee_id,
training_id,
CAST(date_format(completition_date, 'yyyy-MM-dd') AS DATE) AS completition_date,
CAST(date_format(completition_date, 'HH:mm:ss') AS STRING) AS completition_time,
CASE status
        WHEN 1 THEN 'Not Started'
        WHEN 2 THEN 'In Progress'
        else 'Completed'
END as status
FROM zenviro.bronze.employee_training
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.silver.employee_training
AS
SELECT 
employee_id,
training_id,
CAST(date_format(completition_date, 'yyyy-MM-dd') AS DATE) AS completition_date,
CAST(date_format(completition_date, 'HH:mm:ss') AS STRING) AS completition_time,
CASE status
        WHEN 1 THEN 'Not Started'
        WHEN 2 THEN 'In Progress'
        else 'Completed'
END as status
FROM zenviro.bronze.employee_training
```
