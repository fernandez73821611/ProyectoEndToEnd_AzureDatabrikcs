# Transformar Datos Employee Benefits
- Obtener las fechas de los campos start_date y end_date y converti a date
- Escribir los datos transformados en silver

```sql
SELECT *FROM zenviro.bronze.employee_benefits
```

```sql
SELECT 
employee_id,
benefit_id,
CAST(date_format(start_date,"yyyy-MM-dd") AS Date) AS start_date,
CAST(date_format(end_date,"yyyy-MM-dd") AS Date) AS end_date
FROM zenviro.bronze.employee_benefits
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.silver.employee_benefits
AS
SELECT 
employee_id,
benefit_id,
CAST(date_format(start_date,"yyyy-MM-dd") AS Date) AS start_date,
CAST(date_format(end_date,"yyyy-MM-dd") AS Date) AS end_date
FROM zenviro.bronze.employee_benefits
```
