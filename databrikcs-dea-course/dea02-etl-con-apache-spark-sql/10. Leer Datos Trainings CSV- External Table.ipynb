```sql
CREATE TABLE IF NOT EXISTS zenviro.bronze.trainings_01
(
    column_name column_type,
    column_name column_type

)
USING file_format
OPTIONS(
    clave= 'valor'
)
LOCATION 'file_path'
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.bronze.trainings

(
training_id     INT,
training_name   VARCHAR(50),
description     VARCHAR(255),
provider        VARCHAR(50),
duration_hours  INT

)
USING CSV
OPTIONS(
    header='false',
    delimiter=','
)
LOCATION 'abfss://zenviro@adlsjosephdev.dfs.core.windows.net/raw/external-data/trainings'
```
