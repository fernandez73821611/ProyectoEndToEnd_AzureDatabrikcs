```sql
SELECT 
e.employee_id,
e.first_name,
e.last_name,
e.email,
e.phone,
e.status,
j.title,
j.description
FROM zenviro.silver.employees as e
inner join zenviro.silver.job_positions as j
on e.job_position_id=j.job_position_id
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.gold.employee_position
AS
SELECT 
e.employee_id,
e.first_name,
e.last_name,
e.email,
e.phone,
e.status,
j.title,
j.description
FROM zenviro.silver.employees as e
inner join zenviro.silver.job_positions as j
on e.job_position_id=j.job_position_id
```
