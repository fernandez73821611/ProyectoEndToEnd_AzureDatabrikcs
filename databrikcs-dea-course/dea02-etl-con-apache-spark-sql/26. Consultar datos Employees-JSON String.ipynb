# Consultar datos Employees-JSON String
- Extraer valores de columnas de nivel superior
- Extraer elementos de la matriz o arrays
- Extraer valores de columnas anidadas
- Convertir valores de una columnas a un tipo de dato especifico

```sql
SELECT *FROM zenviro.bronze.v_employees limit 10
```

```sql
SELECT 
value:employee_id,
value:department_id,
value
FROM zenviro.bronze.v_employees limit 10
```

```sql
SELECT 
value:employee_id,
value:department_id,
value:name[0].first_name
FROM zenviro.bronze.v_employees limit 10
```

```sql
SELECT 
value:employee_id::INTEGER AS employee_id,
value:birth_date:: DATE AS birth_date,
value:name[0].first_name::STRING AS first_name,
value:name[0].last_name AS last_name,
value
FROM zenviro.bronze.v_employees limit 10
```
