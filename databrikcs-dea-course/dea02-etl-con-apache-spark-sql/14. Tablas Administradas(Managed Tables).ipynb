# Tablas Administradas(Managed Tables)
- Crear una tabla administrada(Managed table)con SQL
- Descibir la Managed Table
- Efectos de eliminar una tabla administrada
- Recuperar una tabla administrada

## 01. Crear una tabla administrada(Managed table)con SQL

```sql
SELECT *FROM zenviro.bronze.v_locations LIMIT 5
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.default.mt_locations
USING DELTA
AS
SELECT *FROM zenviro.bronze.v_locations
```

## 02. Descibir la Managed Table

```sql
SELECT *FROM zenviro.default.mt_locations LIMIT 10
```

```sql
DESCRIBE EXTENDED zenviro.default.mt_locations
```

## 03. Efectos de eliminar una tabla administrada

```sql
DROP TABLE zenviro.default.mt_locations
```

## 04. Recuperar una tabla administrada

```sql
SHOW TABLES DROPPED IN zenviro.default
```

```sql
UNDROP TABLE zenviro.default.mt_locations  -- ESTO SIRVE PARA RESTAURAR LA TABLA ELIMINADA
```
