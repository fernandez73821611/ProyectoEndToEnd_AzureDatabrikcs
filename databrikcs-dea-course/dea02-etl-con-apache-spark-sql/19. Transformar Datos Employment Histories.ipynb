# Transformar Datos Employment Histories
- Extraer la fecha de los campos start_date end_date con los mismos nombres
- Escribir los datos transformados en el schema silver

```sql
SELECT *FROM zenviro.bronze.v_employments_histories
```

```sql
SELECT 
history_id,
employee_id,
department_id,
job_position_id,
CAST(date_format(start_date,"yyyy-MM-dd") AS DATE) AS start_date,
CAST(date_format(end_date,"yyyy-MM-dd") AS DATE) AS end_date,
salary
FROM zenviro.bronze.v_employments_histories
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.silver.employments_histories
AS
SELECT 
history_id,
employee_id,
department_id,
job_position_id,
CAST(date_format(start_date,"yyyy-MM-dd") AS DATE) AS start_date,
CAST(date_format(end_date,"yyyy-MM-dd") AS DATE) AS end_date,
salary
FROM zenviro.bronze.v_employments_histories
```
