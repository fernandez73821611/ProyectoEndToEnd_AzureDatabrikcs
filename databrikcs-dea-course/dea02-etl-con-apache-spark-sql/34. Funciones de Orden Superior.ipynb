#  Funciones de Orden Superior
- Son utilizadas para con datos json array o maps
- TRANSFORM ,FILTER,EXISTS ,AGGREGATE
SINTAXIS:
 nombre_funtion(array, funtion lambda)

```sql
CREATE OR REPLACE TEMPORARY VIEW product_items
AS 
SELECT *FROM
VALUES
    (1,ARRAY('Keyboard','Monitor','Mause','SmartPhone')),
    (2,ARRAY('Tablet','Ram Memory','SmartWatch')),
    (3,ARRAY('Printer','Laptop'))
AS products(product_id,items)
```

```sql
SELECT *FROM product_items
```

```sql
SELECT 
product_id,
TRANSFORM(items,x->UPPER(X)) AS upper_items
FROM product_items
```

```sql
SELECT 
product_id,
FILTER(items,x->x like '%Smart%') AS smart_items
FROM product_items
```

```sql
SELECT 
product_id,
EXISTS(items, x -> x="Tablet") AS has_items
FROM product_items
```

```sql
SELECT 
product_id,
EXISTS(items, x -> x="Tablet") AS has_items
FROM product_items
```

```sql
CREATE OR REPLACE TEMPORARY VIEW product_items
AS
    SELECT * FROM
    VALUES
        (1, array(
                named_struct('name', 'keyboard', 'price', 99),
                named_struct('name', 'monitor', 'price', 899),
                named_struct('name', 'mouse', 'price', 169),
                named_struct('name', 'smartphone', 'price', 799)
        )),
        (2, array(
                named_struct('name', 'tablet', 'price', 359),
                named_struct('name', 'ram memory', 'price', 89),
                named_struct('name', 'smartwatch', 'price', 329)
        )),
        (3, array(
                named_struct('name', 'printer', 'price', 299),
                named_struct('name', 'laptop', 'price', 1299)
        ))
    AS products(product_id, items)
```

```sql
SELECT *FROM product_items
```

```sql
SELECT product_id,
TRANSFORM(items,  x-> named_struct(
    'name',UPPER(x.name),
    'price',ROUND(x.price*1.18,2)
)) AS item_with_tax
FROM product_items
```

```sql
SELECT product_id,
TRANSFORM(items,  x-> named_struct(
    'name',UPPER(x.name),
    'price',ROUND(x.price*1.18,2)
)) AS item_with_tax
FROM product_items
```

```sql
SELECT product_id,
AGGREGATE(items,0,(acc,x)->acc+x.price) as total_product
FROM product_items
```

```sql
SELECT product_id,
EXPLODE(items)
FROM product_items
```
