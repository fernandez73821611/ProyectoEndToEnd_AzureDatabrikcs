# Leer Datos Photochecks -Archivo Binario
- Consultar los elementos de la carpeta Photochecks
- Consultar archivos Photochecks mediante el formato binaryFIle
- Crear una view en el schema bronze

## 01. Consultar los elementos de la carpeta Photochecks

```sql
%fs ls '/Volumes/zenviro/raw/operational_data/photochecks'
```

## 02. Consultar archivos Photochecks mediante el formato binaryFIle

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/photochecks/*/*.png',
    format=>'binaryFile'
)Limit 10
```

## 03. Crear una view en el schema bronze

```sql
CREATE OR REPLACE VIEW zenviro.bronze.v_photochecks 
AS
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/photochecks/*/*.png',
    format=>'binaryFile'
)
```

```sql
SELECT *FROM zenviro.bronze.v_photochecks limit 10;
```
