# Crear acceso a los datos mediante Unity Catalog

- Crear el obejto External Location
- Crear el catalogo zenviro
- Crear los schemas raw , bronze , silver , gold
- Crear el volumen operational-data

## 01. Crear el objeto External Location

```sql
-- Create a location accessed using the abfss_remote_cred credential
CREATE EXTERNAL LOCATION IF NOT EXISTS external_location_zenviro
    URL 'abfss://zenviro@adlsjosephdev.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL storage_credential_adlsjosephdev)
    COMMENT 'Default source for Azure exernal data';
```

## 02. Crear Catalogo zenviro

```sql
CREATE CATALOG IF NOT EXISTS  zenviro
      COMMENT 'Creacion Catalogo zenviro'
      
```

```sql
SELECT current_catalog();
```

```sql
USE CATALOG zenviro;
```

## 03.Crear schemas raw, bronze ,silver, gold

```sql
CREATE SCHEMA IF NOT EXISTS  zenviro.raw;

CREATE SCHEMA IF NOT EXISTS  zenviro.bronze;

CREATE SCHEMA IF NOT EXISTS  zenviro.silver;

CREATE SCHEMA IF NOT EXISTS  zenviro.gold;

     
```

```sql
SELECT current_schema()
```

```sql
USE SCHEMA raw
```

## Crear VOlumen Operational-data

```sql
-- Create an external volume on the specified location with comment
CREATE EXTERNAL VOLUME zenviro.raw.operational_data
    LOCATION 'abfss://zenviro@adlsjosephdev.dfs.core.windows.net/raw/operational-data'
    COMMENT 'Esta es la creacion de mi volumen operational_data'
```
