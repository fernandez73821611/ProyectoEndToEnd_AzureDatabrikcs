# Leer datos JSON en la carpeta Locations
- Consultar Archivos Unicos
- Consultar un conjunto de archivos utilizando comodines
- Consultar todos los archivos de una carpeta
- Seleccionar metadatos de archivos

## 01. Consultar archivos unicos

```sql
SELECT *FROM JSON.`/Volumes/zenviro/raw/operational_data/locations/locations_2025_10.json` limit 5;
```

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/locations/locations_2025_10.json',
    format=>'json'
) limit 10;
```

## 02. Consultar un conjunto de archivos utilizando comodines

```sql
SELECT *FROM JSON.`/Volumes/zenviro/raw/operational_data/locations/locations_2025_*.json`limit 10;
```

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/locations/locations_2025_*.json',
    format=>'json'
) limit 10;
```

## 03. Consultar todos los archivos de una carpeta

```sql
SELECT *FROM JSON.`/Volumes/zenviro/raw/operational_data/locations` limit 5;
```

## 04. Seleccionar metadatos de archivos

```sql
SELECT _metadata.file_name,_metadata.file_size ,*FROM JSON.`/Volumes/zenviro/raw/operational_data/locations` limit 5;
```

```sql
SELECT _metadata.file_path,*FROM read_files(
    '/Volumes/zenviro/raw/operational_data/locations/locations_2025_*.json',
    format=>'json'
) limit 10;
```
