# Registrar Archivos en Unity Catalog usando Views

```sql
CREATE OR REPLACE VIEW zenviro.bronze.v_name
AS
SELECT *FROM read_files(
    ''
)
```

```sql
USE CATALOG zenviro;
USE SCHEMA bronze;
CREATE OR REPLACE VIEW v_name
AS
SELECT *FROM read_files(
    ''
)
```

```sql
CREATE OR REPLACE VIEW zenviro.bronze.v_locations
AS
SELECT _metadata.file_path,*FROM read_files(
    '/Volumes/zenviro/raw/operational_data/locations/'
)
```

```sql
SELECT *FROM zenviro.bronze.v_locations LIMIT 5;
```
