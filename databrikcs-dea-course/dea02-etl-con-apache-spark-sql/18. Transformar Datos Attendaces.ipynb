# Transformar Datos Attendances
- Extraer la fecha del campo check_in en un nuevo campo llamado date
- Extraer la hora de check_in y check_out crear campos check_in_time and check_out_time
- Asignar al campues status los siguients valores:1. Present, 2:Absent, 3:Remote, 4:Late, 5:Vacation
6:On Leave
- Ecribir los datos transformados en en el schema silver

```sql
SELECT *FROM zenviro.bronze.v_attendances limit 10
```

```sql
SELECT 
CAST(attendance_id AS INT)AS attendance_id,
CAST(employee_id AS INT) AS employee_id,
CAST(date_format(check_in,'yyyy-MM-dd') AS date) AS date,
CAST(date_format(check_in,'HH:mm:ss') AS STRING) AS check_in_time,
CAST(date_format(check_out,'HH:mm:ss') AS STRING) AS check_out_time,
CAST(status AS STRING) AS status
FROM zenviro.bronze.v_attendances 
```

```sql
CREATE OR REPLACE TEMPORARY VIEW tv_attendances AS
SELECT 
CAST(attendance_id AS INT)AS attendance_id,
CAST(employee_id AS INT) AS employee_id,
CAST(date_format(check_in,'yyyy-MM-dd') AS date) AS date,
CAST(date_format(check_in,'HH:mm:ss') AS STRING) AS check_in_time,
CAST(date_format(check_out,'HH:mm:ss') AS STRING) AS check_out_time,
CAST(status AS STRING) AS status
FROM zenviro.bronze.v_attendances 
```

```sql
SELECT 
attendance_id,
employee_id,
date,
check_in_time,
check_out_time,
CASE
    WHEN status = 1 THEN "Present"
    WHEN status = 2 THEN "Absent"
    WHEN status = 3 THEN "Remote"
    WHEN status = 4 THEN "Late"
    WHEN status = 5 THEN "Vacation"
    ELSE "On Leave"
END AS status
FROM tv_attendances
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.silver.attendances
AS
SELECT 
attendance_id,
employee_id,
date,
check_in_time,
check_out_time,
CASE
    WHEN status = 1 THEN "Present"
    WHEN status = 2 THEN "Absent"
    WHEN status = 3 THEN "Remote"
    WHEN status = 4 THEN "Late"
    WHEN status = 5 THEN "Vacation"
    ELSE "On Leave"
END AS status
FROM tv_attendances
```

```sql
SELECT *FROM zenviro.silver.attendances limit 10
```
