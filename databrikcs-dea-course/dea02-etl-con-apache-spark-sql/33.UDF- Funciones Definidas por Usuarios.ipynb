# Funciones Definidas por Usuarios -UDF

CREATE OR REPLACE FUNCTION catalog.schema.nombre_functions

RETURNS type_date

RETURN expresion


```sql
SELECT *FROM zenviro.silver.employees limit 5;
```

```sql
CREATE OR REPLACE FUNCTION zenviro.default.get_full_name(first_name STRING,last_name STRING)
RETURNS STRING
RETURN CONCAT(initcap(first_name), ' ', initcap(last_name))
```

```sql
SELECT zenviro.default.get_full_name("Isabel","Sanz")
```

```sql
DESCRIBE FUNCTION EXTENDED zenviro.default.get_full_name
```

```sql
SELECT 
employee_id,
zenviro.default.get_full_name(first_name,last_name) AS full_name
FROM zenviro.silver.employees limit 5;
```
