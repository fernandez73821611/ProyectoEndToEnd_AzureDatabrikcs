# Leer Datos pictures -Archivo Binario
- Consultar los elementos de la carpeta pictures
- Consultar archivos pictures mediante el formato binaryFIle
- Crear una view en el schema bronze

## 01. Consultar los elementos de la carpeta pictures

```sql
%fs ls '/Volumes/zenviro/raw/operational_data/pictures'
```

## 02. Consultar archivos pictures mediante el formato binaryFIle

```sql
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/pictures/*/*.jpg',
    format=>'binaryFile'
)Limit 10
```

## 03. Crear una view en el schema bronze

```sql
CREATE OR REPLACE VIEW zenviro.bronze.v_pictures 
AS
SELECT *FROM read_files(
    '/Volumes/zenviro/raw/operational_data/pictures/*/*.jpg',
    format=>'binaryFile'
)
```

```sql
SELECT *FROM zenviro.bronze.v_pictures limit 10;
```
