# Resumen-Gasto por Ubicacion

```sql
SELECT *FROM zenviro.silver.departments limit 5;
```

```sql
SELECT *FROM zenviro.silver.employees limit 5;
```

```sql
SELECT *FROM zenviro.silver.employments_histories limit 5
```

```sql
SELECT 
l.country,
l.city,
d.department_name,
e.employee_id,
e.first_name,
ei.salary

FROM zenviro.silver.employments_histories AS ei 
       INNER JOIN zenviro.silver.employees AS e 
       ON ei.employee_id = e.employee_id
       INNER JOIN zenviro.silver.departments AS d 
       ON e.department_id = d.department_id
       INNER JOIN zenviro.silver.locations AS l 
       ON d.location_id = l.location_id
      
```

```sql

SELECT 
l.city,
d.department_name,
count(e.employee_id) as total_employee,
SUM(ei.salary) AS total_salary,
ROUND(AVG(ei.salary),2) AS promedio_salary

FROM zenviro.silver.employments_histories AS ei 
       INNER JOIN zenviro.silver.employees AS e 
       ON ei.employee_id = e.employee_id
       INNER JOIN zenviro.silver.departments AS d 
       ON e.department_id = d.department_id
       INNER JOIN zenviro.silver.locations AS l 
       ON d.location_id = l.location_id
GROUP BY l.country,l.city,d.department_name
ORDER BY l.country,l.city,d.department_name
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.gold.employees_by_locations_summary
AS
SELECT 
l.city,
d.department_name,
count(e.employee_id) as total_employee,
SUM(ei.salary) AS total_salary,
ROUND(AVG(ei.salary),2) AS promedio_salary

FROM zenviro.silver.employments_histories AS ei 
       INNER JOIN zenviro.silver.employees AS e 
       ON ei.employee_id = e.employee_id
       INNER JOIN zenviro.silver.departments AS d 
       ON e.department_id = d.department_id
       INNER JOIN zenviro.silver.locations AS l 
       ON d.location_id = l.location_id
GROUP BY l.country,l.city,d.department_name
ORDER BY l.country,l.city,d.department_name
```
