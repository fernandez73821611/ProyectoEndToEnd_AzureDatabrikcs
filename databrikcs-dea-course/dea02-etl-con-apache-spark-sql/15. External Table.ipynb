# Tabla Externa(External Table)

```sql
SELECT 
location_id,
location_name,
country
FROM zenviro.bronze.v_locations limit 10
```

```sql
CREATE TABLE IF NOT EXISTS zenviro.default.ext_table_locations
(
    location_id INT,
    location_name STRING,
    country STRING
)
USING DELTA
LOCATION 'abfss://demo@adlsjosephdev.dfs.core.windows.net/ext_table_locations'

```

```sql
INSERT INTO zenviro.default.ext_table_locations
SELECT 
location_id,
location_name,
country
FROM zenviro.bronze.v_locations 
```

```sql
SELECT *FROM zenviro.default.ext_table_locations LIMIT 10
```

```sql
DESCRIBE EXTENDED zenviro.default.ext_table_locations
```

```sql
DROP TABLE zenviro.default.ext_table_locations
```
