# Leer Datos de Job Partitions - Json Complex a TXT

- Cuando tengas un archivo json complejo en tu capa landing o raw , es mejor ingestar un tipo de dato txt
  Para luego aplicar una logica y tratar con columnas corruptas 

```python
df_job_positions=spark.read.format("json").load("volumes/zenviro/raw/operarional_data/job_positions/")
```

```python
df_job_positions=spark.read.format("txt").load("volumes/zenviro/raw/operarional_data/job_positions/")
```

```python
df_job_positions.writeTo("zenviro.bronze.job_positions_py").createOrReplace()
```
