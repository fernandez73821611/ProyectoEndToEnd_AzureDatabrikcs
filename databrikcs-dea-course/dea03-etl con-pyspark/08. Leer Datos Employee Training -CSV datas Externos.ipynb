# Leer Datos de Employee Traning-CSV(Archivo externo y no tiene cabeceras)

```python
from pyspark.sql.types import StructType, StructField, IntegerType, TimestampType
```

```python
schema=StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("training_id", IntegerType(), True),
    StructField("completion_date", TimestampType(), True),
    StructField("status", IntegerType(), True),
])
```

```python
df_training=(
spark.read.format("csv")
    .option("header", "false")
    .option("delimiter", ",")
    .schema(schema)
    .load("abfss://zenviro@storageaccount.dfs.core.windows.net/raw/external_data/employee_training/")
)
```

```python
df_training=(
spark.read.format("csv")
    .option("header", "false")
    .option("delimiter", ",")
    .schema("employee_id INT, training_id INT, completion_date TIMESTAMP, status INT")
    .load("abfss://zenviro@storage|account.dfs.core.windows.net/raw/external_data/training/")
)
```
