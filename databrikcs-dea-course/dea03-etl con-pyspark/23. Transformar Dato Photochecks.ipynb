# Transformar Datos Photochecks
- Obtener el valor del employee_id de la columna path del archivo
- Escribir los datos transformados en silver

```python
df_photochecks=spark.table("zenviro.bronze.photochecks_py")
```

```python
from pyspark.sql.functions import col, regexp_extract
```

```python
df_photochecks_transform=df_photochecks.select(
    regexp_extract(col("path"), r".*/([0-9]+)\\.png$", 1).cast("int").alias("employee_id"),
    col("content").cast("binary").alias("employee_photocheck")
)
```

```python
df_photochecks_transform.writeTo("zenviro.silver.photochecks_py").createOrReplace()
```
