# Transformar Datos Tranings

- Convertir la primer letra en mayuscula de los campos training_name ,description y provider
- Escribir los datos transformados en el schema silver

```python
df_training=spark.table("zenviro.bronze.training_py")
```

```python
display(df_training)
```

```python
from pyspark.sql.functions import initcap,col
```

```python
df_training_mayus=df_training.select(
    "training_id",
    initcap(col("training_name")).alias("training_name"),
    initcap(col("description")).alias("description"),
    initcap(col("provider")).alias("provider"),
    "duration_hours",
)
```

```python
df_training_mayus.writeTo("zenviro.silver.training_py").createOrReplace()
```
