# Transformar Datos Employee Training

- Extraer fecha y hora del campo completation_date y crear columnas completation_date, completation_time
- Asignar al campus status 1:Not Started , 2: In Progress, 3: Completed
- Escribir los datos transformados en el schema silver

```python
df_employee_training=spark.table("zenviro.bronze.employee_training_py")
```

```python
display(df_employee_training)
```

```python
from pyspark.sql.functions import date_format, col,when
```

```python
df_employee_training_tranform=df_employee_training.select(
    "employee_id",
    "training_id",
    date_format(col("completion_date"),"yyyy-MM-dd").cast("date").alias("completion_date"),
    date_format(col("completion_date"),"HH:mm:ss").cast("string").alias("completion_time"),
    when(col("status")==1,"Not Started")
    .when(col("status")==2,"In Progress")
    .when(col("status")==3,"Completed")
)
```

```python
df_employee_training_transform.writeTo("zenviro.silver.employee_training_py").createOrReplace()
```
