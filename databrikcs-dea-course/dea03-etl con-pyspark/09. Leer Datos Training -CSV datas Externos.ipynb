# Leer Datos de Traning-CSV(Archivo externo y no tiene cabeceras)

```python
from pyspark.sql.types import StructType, StructField, IntegerType, FloatType,StringType
```

```python
schema=StructType([
    StructField("training_id", IntegerType(), True),
    StructField("training_name", StringType(), True),
    StructField("description", StringType(), True),
    StructField("provider", StringType(), True),
    StructField("duration_hours", IntegerType(), True)
])
```

```python
df_training=(
spark.read.format("csv")
    .option("header", "false")
    .option("delimiter", ",")
    .schema(schema)
    .load("abfss://zenviro@storageaccount.dfs.core.windows.net/raw/external_data/training/")
)
```

```python
df_training=(
spark.read.format("csv")
    .option("header", "false")
    .option("delimiter", ",")
    .schema("training_id INT, training_name STRING, description STRING, provider STRING, duration_hours INT")
    .load("abfss://zenviro@storageaccount.dfs.core.windows.net/raw/external_data/training/")
)
```

```python
df_training.writeTo("zenviro.bronze.training_py").createOrReplace()
```
