# Convinar Datos JOIN

```python
df_employees=spark.table("zenviro.silver.employee_py")
```

```python
display(df_employees)
```

```python
df_job_positions=spark.table("zenviro.silver.job_positions_py")
```

```python
df_employees_join=df_employees.join(
    df_job_positions, df_employees.id==df_job_positions.id, "inner"
).select("*")
```
