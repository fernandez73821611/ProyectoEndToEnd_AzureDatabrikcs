# Transformar Datos Benefits
- Sustituir el caracter - por un espacio en el campo description
- Escribir los datos transformados en el schema silver

```python
df_benefits=spark.table("zenviro.bronze.benefits_py")
```

```python
display(df_benefits)
```

```python
from pyspark.sql.functions import col
```

```python
df_benfits_description=df_benefits.select(
    col("benefit_id").cast("int").alias("benefit_id"),
    col("benefit_name").cast("string").alias("benefit_name"),
    replace(col("description"),"-"," ").cast("string").alias("description"),
    col("type").cast("string").alias("type"),
)
```

```python
df_benefits_description.writeTo("zenviro.silver.benefits_py").createOrReplace()
```
