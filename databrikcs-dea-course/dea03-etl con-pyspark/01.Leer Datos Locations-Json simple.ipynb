- Otra manera de consulatar mas flexible

```python
df_locations=spark.read.format("json").load("volumes/zenviro/raw/operarional_data/location") #Utilizando el metodo load para leer multiples archivos json
```

- Agregar Metadatos al pazar a la capa Bronze con _metadata.

```python
df_locations_metada=df_locations.select("_metadata.file_path","*")

display(df_locations_metada)
```

- Para guardar o escribir datos en la capa bronze

```python
df_locations_metada.write.format("delta").mode("overwrite").saveAsTable("zenviro.bronze.locations_py")
```

```python
df_locations_metada.createOrReplaceTable("zenviro.bronze.locations_py")
```

```python
df_locations_metada.writeTo("zenviro.bronze.locations_py").createOrReplace()
```
