# Leer Datos Departments- JSON simple 

```python
df_departments=spark.format("json").load("volumes/zenviro/raw/operarional_data/departments/")
```

```python
display(df_departments.limit(10))
```

```python
df_departments_metadata=df_departments.select("_metadata.file_path","*")
```

```python
df_departments_metadata.writeTo("zenviro.bronze.departments_py").createOrReplace()
```
