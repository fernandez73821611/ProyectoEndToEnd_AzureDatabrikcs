# Transformar Datos Employee Benefits
- Obtener  la fecha de los campos start_date y end_date  y convir a tipo de dato date
- Escribir los datos transformados en el schema silver

```python
df_employee_benefits=spark.table("zenviro.bronze.employee_benefits_py")
```

```python
display(df_employee_benefits)
```

```python
from pyspark.sql.functions import date_format,col
```

```python
df_employee_benefits_date=df_employee_benefits.select(
    "employee_id",
    "benefit_id",
    date_format(col("start_date"),"yyyy-MM-dd").cast("date").alias("start_date"),
    date_format(col("end_date"),"yyyy-MM-dd").cast("date").alias("end_date")   
)
```

```python
df_employee_benefits_date.writeTo("zenviro.silver.employee_benefits_py").createOrReplace()
```
