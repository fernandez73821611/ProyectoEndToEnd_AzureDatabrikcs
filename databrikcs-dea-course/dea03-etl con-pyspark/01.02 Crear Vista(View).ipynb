# Crear una Vista (View)

- Crear una vista temporal 

```python
df.createOrReplaceTempView("tv_locations")
```

- Crear una vista temporal global

```python
df.createOrReplaceGlobalTempView("tvg_locations_global")
```

- Para Crear una vista estandar no existe en pyspark por la cual tienes que utilizar un sparkl.sql
- Se crea una vista estandar a partir de datos de una tabla 
- No se puede crear una vista permanente a partir de una vista temporal 

```python
spark.sql("CREATE OR REPLACE VIEW zenviro.bronze.vw_locations  AS SELECT * FROM tv_locations")
```
